```python
cdf = df['trending_date'].apply(lambda x: '20' + x[:2]).value_counts().to_frame().reset_index()
cdf.columns = ['year', 'no_of_videos']
```

```python
import matplotlib.pyplot as plt

plt.bar(cdf['year'], cdf['no_of_videos'])
plt.xlabel("Year")
plt.ylabel("Number of Videos")
plt.title("Trending Videos per Year")
plt.show()
```
